```python
import tensorflow as tf 
print("tensorflow version" , tf.__version__)
```

```python
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0
```

```python
model = tf.keras.models.Sequential([
  tf.keras.layers.Input(shape=(28, 28)),# flattens the image
  tf.keras.layers.Dense(128, activation='relu'),# hidden layer Wth 128 neurons and connection to 784 input 
  tf.keras.layers.Dropout(0.2),# reduces overfitting probability by 20%
  tf.keras.layers.Dense(10)# output layer with 10 neurons for 10 classes
])
```

```python
predictions = model(x_train[:1]).numpy()
predictions
```

```python
tf.nn.softmax(predictions).numpy()
```

```python
loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True) # loss function to measure the probability error in discrete classification tasks in which the classes are mutually exclusive (each entry is in exactly one class). The loss function takes integer labels and a vector of logits as input.
model = tf.keras.models.Sequential([
    tf.keras.layers.Input(shape=(28, 28)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(10)
])

predictions = model(x_train[:1])
loss_value = loss_fn(y_train[:1], predictions).numpy()
loss_value
```

```python
model.compile(optimizer='adam',
              loss=loss_fn,
              metrics=['accuracy'])
```

```python
model.fit(x_train, y_train, epochs=5)
```

```python
model.fit(x_train, y_train, epochs=5)
```

```python
model.evaluate(x_test,  y_test, verbose=2)
```

```python
model.evaluate(x_test,  y_test, verbose=2)
```

```python
probability_model = tf.keras.Sequential([
  model,
  tf.keras.layers.Softmax()
])
probability_model(x_test[:5])
```
